```python
from itertools import islice

def group_by_3(lst):
    iterator = iter(lst)
    return list(iter(lambda: list(islice(iterator, 3)), []))
completions = group_by_3(completions)
```

```python
func_names = []
test_cases = []
for sample in batch:
    test = sample['test_list'][0]
    func_name = test[7:test.find('(')]
    func_names.append(func_name)
    test_cases.append('\n'.join(sample['test_list']))

# Replace solution with actual function names
completions_with_names = []
for completion_list, func_name in zip(completions, func_names):
    completions_for_prompt = []
    for completion in completion_list:
        completion_with_name = completion.replace('def solution', f'def {func_name}')
        completions_for_prompt.append(completion_with_name)
    completions_with_names.append(completions_for_prompt)
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from datasets import load_dataset
import torch
from activation_pruning import ActivationBasedPruning

def test_pruning_effect():
    # Load model
    print("Loading model...")
    model_name = "Salesforce/codegen-350M-mono"
    model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype='auto', device_map='auto')
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    
    # Initialize pruning
    print("\nInitializing pruning...")
    dataset = load_dataset("mbpp", split="test")
    pruner = ActivationBasedPruning(model, tokenizer, dataset)
    
    # Get initial parameter statistics
    print("\nGathering initial statistics...")
    initial_weights = {}
    for layer_idx, layer in enumerate(model.transformer.h):
        initial_weights[f'layer_{layer_idx}'] = {
            'mlp_in': layer.mlp.fc_in.weight.clone(),
            'mlp_out': layer.mlp.fc_out.weight.clone(),
            'attn_qkv': layer.attn.qkv.weight.clone(),  # Changed this line
            'attn_out': layer.attn.out.weight.clone()   # Added this line
        }
    
    # Apply aggressive pruning
    print("\nApplying pruning...")
    prune_specs = {
        'heads': 64,     # 20% of heads
        'neurons': 800,  # 20% of neurons
        'embeddings': 100
    }
    
    # Compute importance scores and prune
    pruner.importance_scores = pruner.compute_importance_scores()
    pruner.prune_model(prune_specs)
    
    # Check weight changes
    print("\nChecking weight changes...")
    total_params = 0
    zero_params = 0
    changed_params = 0
    
    for layer_idx, layer in enumerate(model.transformer.h):
        current = {
            'mlp_in': layer.mlp.fc_in.weight,
            'mlp_out': layer.mlp.fc_out.weight,
            'attn_qkv': layer.attn.qkv.weight,
            'attn_out': layer.attn.out.weight
        }
        
        for comp_name in current:
            initial = initial_weights[f'layer_{layer_idx}'][comp_name]
            after = current[comp_name]
            
            total_params += initial.numel()
            zero_params += (after == 0).sum().item()
            changed_params += (initial != after).sum().item()
            
            # Print detailed stats for first layer
            if layer_idx == 0:
                print(f"\nLayer 0 {comp_name}:")
                print(f"Total parameters: {initial.numel()}")
                print(f"Zero parameters: {(after == 0).sum().item()}")
                print(f"Changed parameters: {(initial != after).sum().item()}")
                print(f"Max abs difference: {(initial - after).abs().max().item()}")
                
                # Additional check for attention weights
                if comp_name.startswith('attn'):
                    print(f"Shape: {initial.shape}")
                    if torch.allclose(initial, after):
                        print("WARNING: Attention weights unchanged!")
    
    print("\nOverall Statistics:")
    print(f"Total parameters: {total_params:,}")
    print(f"Zero parameters: {zero_params:,} ({zero_params/total_params*100:.2f}%)")
    print(f"Changed parameters: {changed_params:,} ({changed_params/total_params*100:.2f}%)")

if __name__ == "__main__":
    test_pruning_effect()
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from datasets import load_dataset
import torch
from activation_pruning import ActivationBasedPruning

def test_pruning_effect():
    # Load model
    print("Loading model...")
    model_name = "Salesforce/codegen-350M-mono"
    model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype='auto', device_map='auto')
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    
    # Initialize pruning
    print("\nInitializing pruning...")
    dataset = load_dataset("mbpp", split="test")
    pruner = ActivationBasedPruning(model, tokenizer, dataset)
    
    # Get initial parameter statistics
    print("\nGathering initial statistics...")
    initial_weights = {}
    for layer_idx, layer in enumerate(model.transformer.h):
        initial_weights[f'layer_{layer_idx}'] = {
            'mlp_in': layer.mlp.fc_in.weight.clone(),
            'mlp_out': layer.mlp.fc_out.weight.clone(),
            'attn_qkv': layer.attn.qkv_proj.weight.clone(),
            'attn_out': layer.attn.out_proj.weight.clone()
        }
    
    # Print initial shapes
    print("\nInitial shapes (Layer 0):")
    for name, weight in initial_weights['layer_0'].items():
        print(f"{name}: {weight.shape}")
    
    # Apply aggressive pruning
    print("\nApplying pruning...")
    prune_specs = {
        'heads': 64,     # 20% of heads
        'neurons': 800,  # 20% of neurons
        'embeddings': 100
    }
    
    # Compute importance scores and prune
    pruner.importance_scores = pruner.compute_importance_scores()
    pruner.prune_model(prune_specs)
    
    # Check weight changes
    print("\nChecking weight changes...")
    total_params = 0
    zero_params = 0
    changed_params = 0
    
    for layer_idx, layer in enumerate(model.transformer.h):
        current = {
            'mlp_in': layer.mlp.fc_in.weight,
            'mlp_out': layer.mlp.fc_out.weight,
            'attn_qkv': layer.attn.qkv_proj.weight,
            'attn_out': layer.attn.out_proj.weight
        }
        
        for comp_name in current:
            initial = initial_weights[f'layer_{layer_idx}'][comp_name]
            after = current[comp_name]
            
            total_params += initial.numel()
            zero_params += (after == 0).sum().item()
            changed_params += (initial != after).sum().item()
            
            # Print detailed stats for first layer
            if layer_idx == 0:
                print(f"\nLayer 0 {comp_name}:")
                print(f"Shape: {initial.shape}")
                print(f"Total parameters: {initial.numel()}")
                print(f"Zero parameters: {(after == 0).sum().item()}")
                print(f"Changed parameters: {(initial != after).sum().item()}")
                if not torch.allclose(initial, after):
                    print("Max abs difference: {:.6f}".format((initial - after).abs().max().item()))
                else:
                    print("WARNING: Weights unchanged!")
    
    print("\nOverall Statistics:")
    print(f"Total parameters: {total_params:,}")
    print(f"Zero parameters: {zero_params:,} ({zero_params/total_params*100:.2f}%)")
    print(f"Changed parameters: {changed_params:,} ({changed_params/total_params*100:.2f}%)")

if __name__ == "__main__":
    test_pruning_effect()
```
